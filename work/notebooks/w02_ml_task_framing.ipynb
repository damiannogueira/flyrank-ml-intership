## 2. Target or proxy

### Provisional target

The target should represent an observed future outcome for each page, rather than a label created from a fixed rule.

For this lane, a useful provisional target would indicate whether a page shows a meaningful decline in visibility or performance during a later time window. This future outcome could then help determine which pages should have received higher review priority earlier.

I would not use `is_declining_label` as the final target because it is derived from `trend_direction` and `trend_pct`. Using it would make the model learn an existing rule instead of learning from an independent observed outcome.

## 3. Success metric

### Metric: Precision@50

I will use Precision@50 as the main success metric because the practical value of the ranking is concentrated at the top of the list.

Precision@50 measures how many of the 50 highest-priority pages actually show the observed future outcome defined for the target.

For example, a Precision@50 of 0.70 would mean that 35 of the top 50 recommended pages match that outcome.

As a provisional success criterion, I would consider the approach useful if it achieves around 0.70 Precision@50 on held-out future data and performs better than a transparent fixed-rule baseline.

## 5. Why ML beats a fixed rule here

A fixed rule can be useful as a transparent baseline, but it may be too limited for this problem.

Page priority can depend on several signals at the same time, such as impressions, clicks, CTR, average position, freshness, and recent movement. These signals can interact in ways that are difficult to express with a single if-statement.

For example, a page may deserve attention because it is visible but declining, because it is stale and still receives traffic, or because several weaker signals combine into a meaningful pattern.

ML may help by learning these combinations and producing a more flexible ranking, as long as it performs better than a simple rule and remains useful for decision-support.
